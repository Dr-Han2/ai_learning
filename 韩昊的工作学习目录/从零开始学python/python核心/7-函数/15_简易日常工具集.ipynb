# 简易日常工具集
## 1. 场景引入
日常生活中经常需要一些小计算，比如单位换算（米转厘米）、金额计算（商品总价）、时间计算（秒转时分秒）。我们可以把这些功能做成“工具集”，随时调用。
## 2. 核心目标
综合运用“可变参数、多返回值、作用域隔离”等知识点，实现高复用性工具函数。

## 3. 需求拆解与函数规划
1. 函数1：unit_convert(length, unit_from="m", unit_to="cm") → 长度单位换算（默认米转厘米）
2. 函数2：calculate_total(price, *quantities) → 多商品总价（支持买多个相同商品，可变参数）
3. 函数3：seconds_to_hms(total_seconds) → 秒转时分秒（返回小时、分钟、秒，多返回值）
